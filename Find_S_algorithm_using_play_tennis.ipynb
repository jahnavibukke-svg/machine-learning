```python
# Find-S Algorithm
# Dataset: Play Tennis

# Attributes:
# Outlook, Temperature, Humidity, Wind
# Target: Play Tennis Yes (+) / No (-)

data = [
    ["Sunny",    "Hot",  "High",   "Weak",   "-"],
    ["Sunny",    "Hot",  "High",   "Strong", "-"],
    ["Overcast", "Hot",  "High",   "Weak",   "+"],
    ["Rain",     "Mild", "High",   "Weak",   "+"],
    ["Rain",     "Cool", "Normal", "Weak",   "+"],
    ["Rain",     "Cool", "Normal", "Strong", "-"],
    ["Overcast", "Cool", "Normal", "Strong", "+"],
    ["Sunny",    "Mild", "High",   "Weak",   "-"],
    ["Sunny",    "Cool", "Normal", "Weak",   "+"],
    ["Rain",     "Mild", "Normal", "Weak",   "+"],
    ["Sunny",    "Mild", "Normal", "Strong", "+"],
    ["Overcast", "Mild", "High",   "Strong", "+"],
    ["Overcast", "Hot",  "Normal", "Weak",   "+"],
    ["Rain",     "Mild", "High",   "Strong", "-"]
]

# Step 1: Initialize the most specific hypothesis
hypothesis = ["Ø", "Ø", "Ø", "Ø"]

print("Initial Hypothesis:", hypothesis)
print()

# Step 2: Process each training example
for i, example in enumerate(data):

    attributes = example[:-1]
    target = example[-1]

    # Find-S considers ONLY positive examples
    if target == "+":

        # If hypothesis is still completely specific,
        # copy the first positive example
        if hypothesis == ["Ø", "Ø", "Ø", "Ø"]:
            hypothesis = attributes.copy()

        else:
            # Compare each attribute
            for j in range(len(hypothesis)):

                if hypothesis[j] != attributes[j]:
                    hypothesis[j] = "?"

        print(f"After D{i + 1} (+): {hypothesis}")

    else:
        print(f"After D{i + 1} (-): {hypothesis}   (negative example ignored)")

print("\nFinal Most Specific Hypothesis:")
print(hypothesis)
```
