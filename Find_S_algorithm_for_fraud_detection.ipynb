```python
# Find-S Algorithm
# Dataset: E-Commerce Fraud Detection

# Attributes:
# TransactionAmt, ForeignIP, CardPresent
# Target: Fraud (Yes) / Not Fraud (No)

data = [
    ["High", "Yes", "No",  "Yes"],
    ["High", "Yes", "Yes", "No"],
    ["Low",  "Yes", "No",  "No"],
    ["High", "No",  "No",  "No"],
    ["High", "Yes", "No",  "Yes"],
    ["Low",  "No",  "Yes", "No"],
    ["High", "Yes", "No",  "Yes"],
    ["Low",  "Yes", "No",  "No"]
]

# Step 1: Initialize the most specific hypothesis

hypothesis = ["Ø", "Ø", "Ø"]

print("Initial Hypothesis:", hypothesis)
print()

# Step 2: Process each training example

for i, example in enumerate(data):

    attributes = example[:-1]
    target = example[-1]

    # Find-S considers ONLY positive examples

    if target == "Yes":

        print(f"D{i + 1}: Positive Example")

        # If hypothesis is still completely specific,
        # copy the first positive example

        if hypothesis == ["Ø", "Ø", "Ø"]:
            hypothesis = attributes.copy()

        else:
            # Compare each attribute

            for j in range(len(hypothesis)):

                if hypothesis[j] != attributes[j]:
                    hypothesis[j] = "?"

    else:
        print(f"D{i + 1}: Negative Example - Ignored")

    print(f"After D{i + 1}: {hypothesis}")
    print()

# Step 3: Display the final hypothesis

print("--------------------------------")
print("Final Most Specific Hypothesis:")
print(hypothesis)
print("--------------------------------")
```
