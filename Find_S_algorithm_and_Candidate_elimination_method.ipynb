```python
# Find-S Algorithm
# Dataset: Infection Prediction

# Attributes:
# Fever, Cough, Fatigue, Chills
# Target: Infected (+) / Not Infected (-)

data = [
    ["High",     "Severe", "Yes", "Yes",  "+"],
    ["High",     "Severe", "Yes", "No",   "+"],
    ["None",     "Mild",   "No",  "No",   "-"],
    ["High",     "Mild",   "Yes", "No",   "+"],
    ["Moderate", "None",   "No",  "Yes",  "-"]
]

# Step 1: Initialize the most specific hypothesis
hypothesis = ["Ø", "Ø", "Ø", "Ø"]

print("Initial Hypothesis:", hypothesis)
print()

# Step 2: Process each training example
for i, example in enumerate(data):

    attributes = example[:-1]
    target = example[-1]

    # Find-S considers ONLY positive examples
    if target == "+":

        # If hypothesis is still completely specific,
        # copy the first positive example
        if hypothesis == ["Ø", "Ø", "Ø", "Ø"]:
            hypothesis = attributes.copy()

        else:
            # Compare each attribute
            for j in range(len(hypothesis)):

                if hypothesis[j] != attributes[j]:
                    hypothesis[j] = "?"

    print(f"After D{i + 1}: {hypothesis}")

print("\nFinal Most Specific Hypothesis:")
print(hypothesis)
```

```python
# Candidate-Elimination Algorithm
# Dataset: Poisonous / Edible

# Attributes:
# Size, Spots, Color
#
# + = Poisonous
# - = Edible
data = [
    ["Small", "Yes", "Green", "+"],
    ["Large", "Yes", "Green", "+"],
    ["Large", "No",  "Brown", "-"]
]
attributes = ["Size", "Spots", "Color"]
# -----------------------------------------
# Initial Specific Boundary
# -----------------------------------------
S = ["Ø", "Ø", "Ø"]
# -----------------------------------------
# Initial General Boundary
# -----------------------------------------
G = [["?", "?", "?"]]
# Function to check whether a hypothesis
# covers an example
def covers(hypothesis, example):
    for h, x in zip(hypothesis, example):
        if h != "?" and h != "Ø" and h != x:
            return False
    return True
# Function to specialize G
def specialize_g(g, example, S):
    specializations = []
    for i in range(len(attributes)):
        if g[i] == "?":
            # Create a specialized hypothesis
            new_hypothesis = g.copy()
            # Use value from S if it is not Ø
            if S[i] != "Ø" and S[i] != example[i]:
                new_hypothesis[i] = S[i]

                if new_hypothesis not in specializations:
                    specializations.append(new_hypothesis)

    return specializations


# -----------------------------------------
# Candidate Elimination
# -----------------------------------------

print("INITIAL")
print("S =", S)
print("G =", G)
print("-" * 50)


for index, row in enumerate(data):

    example = row[:-1]
    target = row[-1]

    print(f"\nProcessing D{index + 1}: {example} -> {target}")

    # =====================================
    # POSITIVE EXAMPLE
    # =====================================
    if target == "+":

        # Remove G hypotheses that do not cover
        # the positive example
        G = [g for g in G if covers(g, example)]

        # Generalize S
        if S == ["Ø", "Ø", "Ø"]:

            S = example.copy()

        else:

            for i in range(len(attributes)):

                if S[i] != example[i]:
                    S[i] = "?"

    # =====================================
    # NEGATIVE EXAMPLE
    # =====================================
    else:

        # Remove S if it covers negative example
        if covers(S, example):
            print("S covers the negative example.")

        # Specialize G
        new_G = []

        for g in G:

            if covers(g, example):

                specializations = specialize_g(
                    g, example, S
                )

                new_G.extend(specializations)

            else:
                new_G.append(g)

        G = new_G

    # Remove duplicate hypotheses
    unique_G = []

    for g in G:
        if g not in unique_G:
            unique_G.append(g)

    G = unique_G

    print("S =", S)
    print("G =", G)


# -----------------------------------------
# FINAL RESULT
# -----------------------------------------

print("\n" + "=" * 50)
print("FINAL VERSION SPACE BOUNDARIES")
print("=" * 50)

print("\nSpecific Boundary (S):")
print(S)

print("\nGeneral Boundary (G):")

for g in G:
    print(g)
```
